```python
!pip install -q gensim
from gensim import corpora
from gensim.models import LdaModel
docs = [
    ['machine', 'learning', 'ai'],
    ['deep', 'learning', 'neural'],
    ['football', 'cricket', 'sports']
]
d = corpora.Dictionary(docs)
corpus = [d.doc2bow(x) for x in docs]
lda = LdaModel(corpus=corpus, id2word=d, num_topics=2)
for i, topic in lda.print_topics():
    print("Topic", i + 1)
    print(topic)
    print()
```
