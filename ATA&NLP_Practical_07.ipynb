```python
import numpy as np
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, accuracy_score

train_docs = [
    "Python, Java, Git, Docker, and Linux for software development and backend systems",
    "Microservices, cloud containers, APIs, databases, and server deployment",
    "Artificial intelligence, neural networks, machine learning, and computer vision",
    "Hardware processors, GPUs, smartphones, chips, and operating systems",
    "Football tournament, championship match, scoring goals, and extra time",
    "Cricket league, batting, bowling, wickets, and international teams",
    "Quarterback, touchdown pass, game winning play, playoffs, and basketball",
    "Tennis grand slam, sets, players, athlete fitness, and stadium coach",
    "Stock market indices, equity shares, Wall Street, and trading returns",
    "Central bank, interest rates, inflation, treasury bonds, and monetary policy",
    "Investment portfolios, mutual funds, asset management, and risk hedge",
    "Banking liquidity, corporate revenue, quarterly earnings, and financial balance"
]

train_labels = [
    "Technology", "Technology", "Technology", "Technology",
    "Sports", "Sports", "Sports", "Sports",
    "Finance", "Finance", "Finance", "Finance"
]

test_docs = [
    "The quarterback threw a game-winning touchdown pass in the final seconds",
    "Developers use Git and Docker containers to deploy backend microservices",
    "Rising interest rates and banking liquidity affect investment returns"
]

test_labels = ["Sports", "Technology", "Finance"]

model = Pipeline([
    ('vectorizer', CountVectorizer(stop_words='english')),
    ('classifier', MultinomialNB(alpha=0.1))
])

model.fit(train_docs, train_labels)

predictions = model.predict(test_docs)
probabilities = model.predict_proba(test_docs)

print(f"Accuracy: {accuracy_score(test_labels, predictions) * 100:.2f}%\n")
print(classification_report(test_labels, predictions))

for doc, pred, prob in zip(test_docs, predictions, probabilities):
    confidence = np.max(prob) * 100
    print(f"Document   : \"{doc}\"")
    print(f"Prediction : {pred} ({confidence:.2f}%)\n")
```
